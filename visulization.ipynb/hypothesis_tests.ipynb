```python
import numpy as np
from scipy import stats

# Sample delivery times in minutes
delivery_times = np.array([
    28, 32, 31, 29, 35, 30, 27, 33, 31, 34,
    29, 32, 28, 36, 30, 31, 29, 33, 32, 28
])

target_time = 30
alpha = 0.05

# One-sample t-test
result = stats.ttest_1samp(
    delivery_times,
    popmean=target_time
)

print("Sample mean:", round(np.mean(delivery_times), 2))
print("Target mean:", target_time)
print("T-statistic:", round(result.statistic, 4))
print("P-value:", round(result.pvalue, 4))
print("95% confidence interval:", result.confidence_interval())

if result.pvalue < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```
