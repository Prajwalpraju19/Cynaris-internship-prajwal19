```python
import pandas as pd

customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "customer_name": ["Prajwal", "Rahul", "Anil", "Sneha"],
    "region": ["South", "North", "East", "West"]
})

sales = pd.DataFrame({
    "customer_id": [1, 2, 3, 5],
    "product": ["Laptop", "Phone", "Tablet", "Monitor"],
    "sales": [50000, 30000, 20000, 15000]
})
#inner join
inner_join = pd.merge(customers, sales, on="customer_id", how="inner")
print(inner_join)

#left join
left_join = pd.merge(customers, sales, on="customer_id", how="left")
print(left_join)

#right join
right_join = pd.merge(customers, sales, on="customer_id", how="right")
print(right_join)

#outer join
outer_join = pd.merge(customers, sales, on="customer_id", how="outer")
print(outer_join)

#pivot table -sales by region and product
data = pd.DataFrame({
    "region": ["South", "South", "North", "North", "East", "East", "West"],
    "product": ["Laptop", "Phone", "Laptop", "Phone", "Laptop", "Phone", "Laptop"],
    "sales": [50000, 30000, 45000, 25000, 40000, 20000, 35000]
})

pivot = pd.pivot_table(
    data,
    values="sales",
    index="region",
    columns="product",
    aggfunc="sum",
    fill_value=0
)

print(pivot)

#pd.(melt)
wide_data = pd.DataFrame({
    "Region": ["South", "North", "East"],
    "Laptop": [50000, 45000, 40000],
    "Phone": [30000, 25000, 20000],
    "Tablet": [20000, 18000, 15000]
})

long_data = pd.melt(
    wide_data,
    id_vars=["Region"],
    var_name="Product",
    value_name="Sales"
)

print(long_data)

#method chaining
result = (
    data
    .query("sales > 20000")
    .assign(sales_with_tax=lambda df: df["sales"] * 1.18)
    .groupby("region", as_index=False)
    .agg(
        total_sales=("sales", "sum"),
        average_sales=("sales", "mean"),
        total_with_tax=("sales_with_tax", "sum")
    )
    .sort_values("total_sales", ascending=False)
)

print(result)
```
