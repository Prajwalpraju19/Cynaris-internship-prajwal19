```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create the dataset
data = {
    "Group A": [55, 60, 65, 65, 70, 70, 70, 75, 80, 90],
    "Group B": [65, 68, 70, 72, 75, 78, 80, 82, 85, 95]
}

df = pd.DataFrame(data)

# Save dataset
df.to_csv("student_scores.csv", index=False)

# Calculate descriptive statistics
for group in df.columns:
    scores = df[group]

    print(f"\n--- {group} ---")
    print("Mean:", scores.mean())
    print("Median:", scores.median())
    print("Mode:", scores.mode().tolist())
    print("Population Variance:", scores.var(ddof=0))
    print("Population Standard Deviation:", scores.std(ddof=0))
    print("25th Percentile (Q1):", scores.quantile(0.25))
    print("50th Percentile (Q2):", scores.quantile(0.50))
    print("75th Percentile (Q3):", scores.quantile(0.75))
    print("90th Percentile:", scores.quantile(0.90))
    print("IQR:", scores.quantile(0.75) - scores.quantile(0.25))

# Frequency distribution
plt.figure(figsize=(8, 5))
plt.hist(
    [df["Group A"], df["Group B"]],
    bins=[50, 60, 70, 80, 90, 100],
    label=["Group A", "Group B"],
    alpha=0.7,
    edgecolor="black"
)
plt.title("Frequency Distribution of Student Scores")
plt.xlabel("Exam Score")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.savefig("frequency_distribution.png")
plt.show()

# Side-by-side box plots
plt.figure(figsize=(8, 5))
plt.boxplot(
    [df["Group A"], df["Group B"]],
    tick_labels=["Group A", "Group B"]
)
plt.title("Comparison of Student Score Distributions")
plt.ylabel("Exam Score")
plt.tight_layout()
plt.savefig("box_plot_comparison.png")
plt.show()

print("\nAnalysis completed successfully.")
```
